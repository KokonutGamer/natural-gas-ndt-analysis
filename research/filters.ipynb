```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread("../images/0081.tif", cv2.IMREAD_GRAYSCALE)
assert image is not None

h, w = image.shape[:2]
center_r, center_c = h // 2, w // 2
half_size = min(h, w) // 2
image = image[
    center_r - half_size : center_r + half_size,
    center_c - half_size : center_c + half_size,
]
```

```python
assert image is not None
median_image = cv2.medianBlur(image, 11)
plt.imshow(median_image, cmap='gray')
plt.tight_layout()
plt.show()
```

```python
assert image is not None
gaussian_image = cv2.GaussianBlur(image, (0, 0), 10)
plt.imshow(gaussian_image, cmap='gray')
plt.tight_layout()
plt.show()
```

```python
assert image is not None
average_image = cv2.blur(image, (5, 5))
plt.imshow(average_image, cmap='gray')
plt.tight_layout()
plt.show()
```
